```python
!pip install faker
import boto3
import json
import uuid
import random
from datetime import datetime,timedelta
from faker import Faker
```

```python
s3=boto3.client(
    "s3",
    aws_access_key_id=AWS_ACCESS_KEY,
    aws_secret_access_key=AWS_SECRET_KEY
)
```

```python
fake=Faker()
def generate_customer():
    return{
        "customer_id":fake.uuid4(),
        "name":fake.name(),
        "email":fake.email(),
        "country":fake.country(),
        "age":random.randint(18,80),
        "signup_date":fake.date_between(start_date="-3y",end_date="today").isoformat()
    }

def write_customers_to_s3(batch_size):
    now=datetime.utcnow()+timedelta(hours=5,minutes=30)
    customers=[generate_customer() for _ in range(batch_size)]
    key=f"raw/customers/customers_data.json"
    body="\n".join(json.dumps(c) for c in customers)
    s3.put_object(Bucket=BUCKET_NAME,Key=key,Body=body)
    print("Customers data generated successfully!!")
```

```python
write_customers_to_s3(100000)
```
